```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
heights = [
    150, 152, 155, 157, 158,
    160, 160, 161, 162, 163,
    164, 165, 165, 166, 167,
    168, 170, 172, 175, 178
]

df = pd.Series(heights)
print('Mean\n', df.mean())
print('Median\n', df.median())

sns.histplot(df, kde=True)
plt.title('Normal Distribution')
plt.xlabel('Heights')
```

```python
salary = [
    20, 22, 24, 25, 26,
    27, 28, 29, 30, 32,
    35, 40, 45, 60, 90
]

df = pd.Series(salary)
print('Mean :', df.mean())
print('Median :', df.median())

sns.histplot(df, kde=True)
plt.title('Right-Skewed Distribution')
plt.xlabel('Salary')
```

```python
marks = [
    30, 45, 60, 70,
    75, 78, 80, 82,
    84, 85, 86, 88,
    90, 92, 95
    ]

df = pd.Series(marks)
print('Mean :', df.mean())
print('Median :', df.median())

sns.histplot(df, color='gold', kde=True)
plt.title('Left-Skewed Distribution')
plt.xlabel('Marks')
```

```python
ages = [
    18, 18, 19, 19, 20,
    20, 20, 21, 21, 22,
    23, 24
]

df = pd.DataFrame({'Age': ages})
sns.histplot(df['Age'], color='Green', kde=True)
plt.title("Unimodal Distribution")
```

```python
ages = [
    18, 18, 19, 19, 20,
    21, 35, 36, 37, 38,
    39, 40
]

df = pd.DataFrame({'Age': ages})
sns.histplot(df['Age'], color='Red', kde=True, bins=6)
plt.title("Bimodal Distribution")
```

```python
ratings = [
    1, 1, 2, 2,
    5, 5, 6, 6,
    9, 9, 10, 10
]

df = pd.DataFrame({'Ratings': ratings})
sns.histplot(df['Ratings'], color='Orange', kde=True)
plt.title("Multimodal Distribution")
```
