```python
import pandas as pd

# Create dataset
data = {
    "Student": ["A","B","C","D","E","F","G","H","I","J"],
    "Passed": ["Yes","No","Yes","Yes","No",
               "Yes","Yes","No","Yes","Yes"]
}

df = pd.DataFrame(data)

df
```

```python
# Total count
total = len(df)
total
```

```python
passed = len(df[df['Passed']=='Yes'])
passed
```

```python
prob_passed = passed/total
print("Probability of passed", prob_passed)
```

```python
failed = len(df[df['Passed']=='No'])
failed
```

```python
prob_failed = failed/total
print("Probability of failed", prob_failed)
```

```python
import pandas as pd

# Dataset
data = {
    "Student": ["A","B","C","D","E","F"],
    "Gender": ["Boy","Boy","Boy","Girl","Girl","Girl"],
    "Passed": ["Yes","No","Yes","Yes","No","Yes"]
}

df = pd.DataFrame(data)

df
```

```python
total = len(df)
total
```

## Probability of Boy and Passed

```python
boys = len(df[df['Gender']=='Boy'])
boys
```

```python
passed = len(df[df['Passed']=='Yes'])
passed
```

```python
boy_passed = len(df[(df['Gender']=='Boy') & (df['Passed']=='Yes')])
boy_passed
```

```python
prob_boy_passed = (boys + passed - boy_passed)/total
prob_boy_passed
```

```python
girl = len(df[df['Gender']=='Girl'])
girl
```

```python
girl_passed = len(df[(df['Gender']=='Girl') & (df['Passed']=='Yes')])
girl_passed
```

```python
prob_girl_passed = (girl + passed - girl_passed)/total
prob_girl_passed
```

```python
failed = len(df[df['Passed']=='No'])
failed
```

```python
girl_failed = len(df[(df['Gender']=='Girl') & (df['Passed']=='No')])
girl_failed
```

```python
prob_girl_failed = (girl + failed - girl_failed)/total
prob_girl_failed
```

```python
prob_girl_failed = len(df[(df['Gender' ] == 'Girl') & (df['Passed' ] == 'No' ) ])/total
prob_girl_failed
```

```python
prob_boy_passed = len(df[(df['Gender' ] == 'Boy') & (df['Passed' ] == 'Yes' ) ])/total
prob_boy_passed
```
