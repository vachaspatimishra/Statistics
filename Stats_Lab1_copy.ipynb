**Exercise 2:** Draw a **Simple Random Sample** of 40 rows (without replacement) and compute the sample mean salary.

```python
sample_rows = df.sample(n=40, random_state=42, replace=False)
sample_rows['salary'].mean()
```

**Exercise 3:** Perform **Stratified Sampling** by `department`: sample 10% from each stratum (at least 2 per department). Compute the combined sample mean salary.

```python
def stratified_sampling(group):
    sample_size = int(len(group)*0.10)
    return group.sample(sample_size, random_state=2)

stratified_sample = df.groupby('department').apply(stratified_sampling)
stratified_sample['salary'].mean()
```

**Exercise 4:** Create a **Cluster Sample** by `region`: randomly pick **1 region** and include all rows from that region. Compute its mean monthly_sales.

**Exercise 1:** Compute `salary` mean and median with all rows.

**Exercise 2:** Remove the **top 1%** salaries and recompute mean & median. Comment on the change.

```python
sort_df = df.sort_values(by='salary', ascending=False)
threshold = sort_df['salary'].quantile(0.99)
filtered_df = sort_df[sort_df['salary']<threshold]
print(filtered_df['salary'].mean())
print(filtered_df['salary'].median())
```

**Exercise 3:** Compute a **10% trimmed mean** for `salary` (drop lowest 5% and highest 5%).

```python
upper_threshold = sort_df['salary'].quantile(0.95)
lower_threshold = sort_df['salary'].quantile(0.05)
filtered_df = sort_df[(sort_df['salary']>=lower_threshold) & (sort_df['salary'] <= upper_threshold)]
print(filtered_df['salary'].mean())
print(filtered_df['salary'].median())
```

# Dispersion: Range, Variance/Std (Population vs Sample), MAD, CV

**Exercise 1:** Compute the **range** of `salary`.

```python
df['salary'].max() - df['salary'].min()
```

**Exercise 2:** Compute **population variance** and **population std dev** of `salary`.

```python
print(df['salary'].var(ddof=0))
print(df['salary'].std(ddof=0))
```

**Exercise 3:** Compute **sample variance** and **sample std dev** of `salary`.

```python
print(df['salary'].var(ddof=1))
print(df['salary'].std(ddof=1))
```

**Exercise 4:** Compute **Mean Absolute Deviation (MAD)** around the mean for `salary`.

```python
mean_salary = df['salary'].mean()
np.mean(abs(df['salary']-mean_salary))
```

**Exercise 5:** Compute **Coefficient of Variation (CV)** for `salary` and `monthly_sales` (population std / mean).

```python
print(df['salary'].std(ddof=0)/df['salary'].mean())
print(df['monthly_sales'].std(ddof=0)/df['monthly_sales'].mean())
```

**Exercise 6:** By **department**, compute sample std dev and CV of `salary`.

```python
print("---department-wise std. deviation---")
print(df.groupby('department')['salary'].std(ddof=1))
print("---department-wise CV---")
print(df.groupby('department')['salary'].std(ddof=1)/df.groupby('department') ['salary'].mean())
```
