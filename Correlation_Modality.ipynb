```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

heights = [
    150, 152, 155, 157, 158,
    160, 160, 161, 162, 163,
    164, 165, 165, 166, 167,
    168, 170, 172, 175, 178
]

df = pd.DataFrame({'Height': heights})

sns.histplot(df['Height'], kde=True)

plt.title("Normal Distribution")
plt.show()
```

```python
print(df['Height'].mean())
print(df['Height'].median())
```

```python
salary = [
    20, 22, 24, 25, 26,
    27, 28, 29, 30, 32,
    35, 40, 45, 60, 90
]

df = pd.DataFrame({'Salary': salary})

sns.histplot(df['Salary'], kde=True)

plt.title("Right Skewed Distribution")
plt.show()
```

```python
print(df['Salary'].mean())
print(df['Salary'].median())
```

```python
marks = [
    30, 45, 60, 70,
    75, 78, 80, 82,
    84, 85, 86, 88,
    90, 92, 95
]

df = pd.DataFrame({'Marks': marks})

sns.histplot(df['Marks'], kde=True)

plt.title("Left Skewed Distribution")
plt.show()
```

```python
print(df['Marks'].mean())
print(df['Marks'].median())
```

```python
ages = [
    18, 18, 19, 19, 20,
    20, 20, 21, 21, 22,
    23, 24
]

df = pd.DataFrame({'Age': ages})

sns.histplot(df['Age'], kde=True)

plt.title("Unimodal Distribution")
plt.show()
```

```python
ages = [
    18, 18, 19, 19, 20,
    21, 35, 36, 37, 38,
    39, 40
]

df = pd.DataFrame({'Age': ages})

sns.histplot(df['Age'], kde=True, bins=6)

plt.title("Bimodal Distribution")
plt.show()
```

```python
ratings = [
    1, 1, 2, 2,
    5, 5, 6, 6,
    9, 9, 10, 10
]

df = pd.DataFrame({'Ratings': ratings})

sns.histplot(df['Ratings'], kde=True)

plt.title("Multimodal Distribution")
plt.show()
```

```python
import pandas as pd

data = {
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [35, 45, 55, 70, 85]
}

df = pd.DataFrame(data)

print(df)

print("\nCovariance:")
print(df.cov())

print("\nCorrelation:")
print(df.corr())
```

```python
import pandas as pd

data = {
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Likes_Python': ['Yes', 'Yes', 'No', 'Yes', 'No', 'No']
}

df = pd.DataFrame(data)

print(df)
```

```python
cross_tab = pd.crosstab(df['Gender'], df['Likes_Python'])

print(cross_tab)
```

```python
cross_tab.plot(kind='bar')
```
