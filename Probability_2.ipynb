## Conditional probability & Bayes Theorem

```python
import pandas as pd

data = {
    "Student": ["A","B","C","D","E","F","G","H","I","J"],
    "Cricket": ["Yes","Yes","Yes","No","No","Yes","No","Yes","No","Yes"],
    "Passed": ["Yes","Yes","No","Yes","No","Yes","No","Yes","Yes","No"]
}

df = pd.DataFrame(data)
df
```

```python
# Total
total = len(df)
total
```

```python
# Play cricket
cricket = len(df[df['Cricket']=='Yes'])
cricket
```

```python
# Passed
passed = len(df[df['Passed']=='Yes'])
passed
```

```python
# Probability of plays cricket
prob_cricket = cricket/total
prob_cricket
```

```python
# Prob. of passed
prob_passed = passed/total
prob_passed
```

```python
# Prob. of plays cricket AND passed
prob_cric_pass = len(df[(df['Cricket' ] == 'Yes') & (df['Passed' ] == 'Yes' )])/len(df)
prob_cric_pass
```

```python
# Prob. of plays cricket OR passed
prob_cric_or_pass = len(df[(df['Cricket' ] == 'Yes') | (df['Passed' ] == 'Yes' )])/len(df)
prob_cric_or_pass
```

### Probability of passed given he plays cricket

```python
passed_given_cricket = prob_cric_pass/prob_cricket
passed_given_cricket
```

### Bayes Theorem - probability of cricket given passed

```python
cricket_given_passed = passed_given_cricket * prob_cricket / prob_passed
cricket_given_passed
```

## Mean Absolute Deviation 

```python
data = {
    "Students" : ["A", "B", "C", "D", "E"],
    "Study_hours" : [2,4,6,8,10]
}

df = pd.DataFrame(data)
df
```

```python
mean_value = df["Study_hours"].mean()
mean_value
```

```python
df["Deviation" ] = df[ "Study_hours"] - mean_value
df
```

```python
df["Absolute_Deviation"] = abs(df["Deviation"])
df
```

```python
mean_absolute_deviation = df["Absolute_Deviation"].mean()
mean_absolute_deviation
```
