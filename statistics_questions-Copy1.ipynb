## Dataset A
`5, 7, 7, 10, 12, 15, 18, 20, 22, 25`


## Part 1: Mean, Median, Mode
1. Find the mean of Dataset A.
2. Find the median of Dataset A.
3. Find the mode of Dataset A.
4. If 100 is added to Dataset A, what happens to the mean?
5. If 100 is added, what happens to the median?


```python
import pandas as pd
a = pd.Series([5, 7, 7, 10, 12, 15, 18, 20, 22, 25])
print(a.mean())
print(a.median())
print(a.mode())
print((a+100).mean())
print((a+100).median())
```

## Part 2: Variance & Standard Deviation
6. Calculate the variance of Dataset A.
7. Calculate the standard deviation of Dataset A.

## Dataset B
`10, 10, 10, 10, 10`

8. Find mean, variance, and standard deviation of Dataset B.
9. Compare spread of Dataset A and Dataset B.


```python
print(a.var())
print(a.std())
```

```python
b = pd.Series([10, 10, 10, 10, 10])
print(b.mean())
print(b.var())
print(b.std())
```

## Part 3: Percentiles & Quartiles
10. Find the 25th percentile of Dataset A.
11. Find the 50th percentile of Dataset A.
12. Find the 75th percentile of Dataset A.
13. Find Q1, Q2, Q3 for Dataset A.
14. Identify values below Q1 and above Q3.


```python
q1 = a.quantile(0.25)
q2 = a.quantile(0.5)
q3 = a.quantile(0.75)
print(q1)
print(q2)
print(q3)

below_q1 = a[a<=q1]
print(below_q1)
above_q3 = a[a>=q3]
print(above_q3)
```

## Part 4: IQR
15. Calculate the IQR of Dataset A.
16. Find the lower bound using IQR.
17. Find the upper bound using IQR.


```python
iqr = q3 - q1
print(iqr)
```

```python
lower_bound = q1 - 1.5*(iqr)
upper_bound = q3 + 1.5*(iqr)
print(lower_bound)
print(upper_bound)
```
